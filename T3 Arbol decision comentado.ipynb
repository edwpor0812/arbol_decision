```python
predictors_col = ["fixed acidity", "residual sugar", "alcohol", "quality"]
target_col = ['quality']
```

```python
predictors = data[predictors_col]
target = data [target_col]
```

```python
predictors
```

```python
# En esta predicción se evaluan segun su acidez fija, su azucar residual, y su alcohol todo para predecir su calidad
```

```python
target
```

```python
# El dataset de la variable objetivo contiene con valores discretos principalmente en 5 y 6.
```

```python
corr = predictors.corr()
corr
```

```python
# Este análisis de matriz de correlaciones muestra que el contenido de alcohol presenta mayor correlación positiva con la calidad del vino (0.476),
```

```python
# indicando que a mayor porcentaje de alcohol, generalmente se obtiene una mejor puntuación de calidad del vino
```

```python
# Por el contrario, la acidez fija (0.124) y el azúcar residual (0.014) tienen correlaciones muy bajas con la calidad,
```

```python
# lo que sugiere que su influencia directa en la predicción de calidad es limitada.
```

```python
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title('Matriz de Correlación')
plt.show()
```

```python
#  El presente análisis asegura que el alcohol será un predictor relevante para los modelos de clasificación,
```

```python
X_train, X_test, y_train, y_test = train_test_split(predictors, target, test_size=0.2, random_state=13)
```

```python
# Se realiza una partición de los datos en conjuntos de entrenamiento y prueba, destinando 80% y el 20% a las observaciones al entrenamiento del modelo.
```

```python
tree = DecisionTreeClassifier()
```

```python
arbol = tree.fit(X_train, y_train)
```

```python
plot_tree(arbol)
```

```python
# El árbol comienza dividiendo según si la calidad es menor o igual a 5.5, y a medida que avanza, las divisiones sucesivas refinan las predicciones.
```

```python
predicciones = arbol.predict(X_test)
```

```python
predicciones
```

```python
# Aqui se está prediciendo una variedad de valores de Calidad (de 3 a 8).
```

```python
# Examinamos la matriz de confusión
pd.crosstab(np.array([y[0]for y in y_test.values.tolist()]), predicciones, rownames=["Actual"], colnames=['Predicciones'])
```

```python
#  Las clases 5 y 6, que son las más predominantes en el conjunto de prueba.
```

```python
accuracy = accuracy_score(y_test,predicciones)
accuracy
```

```python
# De acuerdo con el modelo, ha realizado una predicción perfecta sobre los datos de prueba
```

```python
precision = precision_score(y_test, predicciones, average= 'weighted')
precision
```

```python
# Las predicciones realizadas en el conjunto de prueba fueron correctas, sin errores de clasificación
```

```python
# Este resultado, junto con el valor de accuracy, sugiere que el árbol de decisión presenta un ajuste completo sobre los datos de evaluación.
```

```python
recall = recall_score(y_test, predicciones, average= 'weighted')
recall
```

```python
# Con este resultado se confirma que el árbol de decisión no presentó errores de omisión en la predicción de las distintas categorías.
```

```python
f1 = f1_score(y_test, predicciones, average= 'weighted')
f1
```

```python
#  El árbol de decisión clasifica de manera óptima, sin comprometer ni la cantidad de aciertos ni la cobertura de las clases.
```
